```python
features = ["Age", "Income", "Total_Spend", "NumWebPurchases", "NumStorePurchases", "NumWebVisitsMonth", "Recency"]
```

```python
X = df[features].copy()
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.cluster import KMeans, DBSCAN
from sklearn.neighbors import NearestNeighbors
```

```python
wcss = []
```

```python
for i in range(2, 11):
    kmeans = KMeans(n_clusters = i)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(2, 11), wcss,marker = "o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=6, random_state=42)
df["cluster"] = kmeans.fit_predict(X_scaled)
```

```python
df.head()
```

```python
cluster_summer = df.groupby("cluster")[features].mean()
```

```python
cluster_summer
```

```python
df["cluster"].value_counts()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
pca_data = pca.fit_transform(X_scaled)
df["PCA1"], df["PCA2"], df["PCA3"] = pca_data[:,0], pca_data[:,1], pca_data[:,2]
```

```python
pca_data
```

```python
import plotly.express as px

fig = px.scatter_3d(
    df, x="PCA1", y="PCA2", z="PCA3",
    color=df["cluster"].astype(str),
    title="Customer Segmentation (K-Means) — 3D View",
    labels={"color": "Cluster"},
    opacity=0.7
)
fig.show()
```

```python
# Find eps for DBSCAN using k-distance elbow plot
neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit = neighbors.fit(X_scaled)
distances, indices = neighbors_fit.kneighbors(X_scaled)
distances = np.sort(distances[:, 4])

plt.plot(distances)
plt.title("K-Distance Graph (for choosing DBSCAN eps)")
plt.xlabel("Points sorted by distance")
plt.ylabel("5th nearest neighbor distance")
plt.show()
```

```python
dbscan = DBSCAN(eps=1.2, min_samples=5)   # replace 1.2 with the elbow value you found above
df["cluster_dbscan"] = dbscan.fit_predict(X_scaled)

df["cluster_dbscan"].value_counts()
```

```python
fig2 = px.scatter_3d(
    df, x="PCA1", y="PCA2", z="PCA3",
    color=df["cluster_dbscan"].astype(str),
    title="Customer Segmentation (DBSCAN) — 3D View",
    labels={"color": "Cluster"},
    opacity=0.7
)
fig2.show()
```
